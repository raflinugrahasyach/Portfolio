```python
import networkx as nx

def degree_centrality(G):
    """
    Hitung degree centrality untuk setiap node dalam graph G.
    Degree centrality mengukur jumlah hubungan yang dimiliki oleh node.

    Args:
        G (networkx.Graph): Graf yang ingin dihitung centrality-nya.

    Returns:
        dict: Dictionary yang berisi degree centrality untuk setiap node.
    """
    return nx.degree_centrality(G)


def betweenness_centrality(G):
    """
    Hitung betweenness centrality untuk setiap node dalam graph G.
    Betweenness centrality mengukur seberapa sering suatu node berada di jalur
    pendekest antar dua node lainnya.

    Args:
        G (networkx.Graph): Graf yang ingin dihitung centrality-nya.

    Returns:
        dict: Dictionary yang berisi betweenness centrality untuk setiap node.
    """
    return nx.betweenness_centrality(G)


def closeness_centrality(G, wf_improved=True):
    """
    Hitung closeness centrality untuk setiap node dalam graph G.
    Closeness centrality mengukur seberapa dekat suatu node ke semua node lainnya.

    Args:
        G (networkx.Graph): Graf yang ingin dihitung centrality-nya.
        wf_improved (bool): Jika True, menggunakan formula improved oleh Wasserman dan Faust.

    Returns:
        dict: Dictionary yang berisi closeness centrality untuk setiap node.
    """
    return nx.closeness_centrality(G, wf_improved=wf_improved)


def eigenvector_centrality(G, max_iter=100, tol=1e-06, nstart=None, weight=None):
    """
    Hitung eigenvector centrality untuk setiap node dalam graph G.
    Eigenvector centrality mengukur pengaruh suatu node berdasarkan pengaruh
    node lainnya yang terhubung langsung.

    Args:
        G (networkx.Graph): Graf yang ingin dihitung centrality-nya.
        max_iter (int): Jumlah iterasi maksimum untuk power iteration.
        tol (float): Toleransi error untuk konvergensi power iteration.
        nstart (dict): Nilai awal untuk power iteration.
        weight (str): Nama atribut edge yang digunakan untuk menghitung berat.

    Returns:
        dict: Dictionary yang berisi eigenvector centrality untuk setiap node.
    """
    return nx.eigenvector_centrality(G, max_iter=max_iter, tol=tol, nstart=nstart, weight=weight)


# Contoh penggunaan:
if __name__ == "__main__":
    # Membuat graf contoh
    G = nx.Graph([(0, 1), (0, 2), (0, 3), (1, 2), (1, 3)])

    # Menghitung degree centrality
    degree_centrality_result = degree_centrality(G)
    print("Degree Centrality:", degree_centrality_result)

    # Menghitung betweenness centrality
    betweenness_centrality_result = betweenness_centrality(G)
    print("Betweenness Centrality:", betweenness_centrality_result)

    # Menghitung closeness centrality
    closeness_centrality_result = closeness_centrality(G)
    print("Closeness Centrality:", closeness_centrality_result)

    # Menghitung eigenvector centrality
    eigenvector_centrality_result = eigenvector_centrality(G)
    print("Eigenvector Centrality:", eigenvector_centrality_result)
```

```python
# Cell X: Improved Degree Centrality Visualization – Top 150
import matplotlib.pyplot as plt
import networkx as nx

# 1. Salin & bersihkan graph
G_clean = G.copy()
G_clean.remove_edges_from(nx.selfloop_edges(G_clean))

# 2. Hitung degree centrality & pilih top N
degree_cent = nx.degree_centrality(G_clean)
top_n       = 150
top_nodes   = sorted(degree_cent, key=degree_cent.get, reverse=True)[:top_n]

# 3. Buat subgraph hanya top nodes
G_deg = G_clean.subgraph(top_nodes).copy()

# 4. Compute layout dengan k lebih besar
pos_deg = nx.spring_layout(G_deg, k=1.0, iterations=200, seed=42)
# (alternatif: pos_deg = nx.kamada_kawai_layout(G_deg))

# 5. Ukuran figure + node
plt.figure(figsize=(20,20))
sizes_deg  = [degree_cent[n]*3000 for n in G_deg.nodes()]
colors_deg = [degree_cent[n]      for n in G_deg.nodes()]

# 6. Gambar nodes & edges
nx.draw_networkx_nodes(G_deg, pos_deg,
                       node_size=sizes_deg,
                       node_color=colors_deg,
                       cmap=plt.cm.Blues,
                       alpha=0.9)
nx.draw_networkx_edges(G_deg, pos_deg,
                       alpha=0.4, width=1)

# 7. Hanya label top-30 agar tidak numpuk
label_count = 30
labels_deg = {n: n for n in top_nodes[:label_count]}
nx.draw_networkx_labels(G_deg, pos_deg,
                        labels=labels_deg,
                        font_size=12,
                        font_color="black")

plt.title('Degree Centrality – Top 150', fontsize=20)
plt.axis('off')
plt.tight_layout()
plt.show()
```
